```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D 
```

```python
expression_coords = np.load('Data_Folder/expression_coords.npy')
expression_data = np.load('Data_Folder/expression.npy',
                          allow_pickle=True).item()
```

```python
#46 genes included
print(expression_data.keys())
```

```python
#plot the expression data
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(expression_coords[:,0], expression_coords[:,1], expression_coords[:,2],
           c=expression_data['GRIA1'], cmap='turbo')
```
