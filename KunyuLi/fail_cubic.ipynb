```python
from scipy.interpolate import griddata
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D


samples = np.load('coords.npy')
band_powers = np.load('band_powers.npy',allow_pickle=True).item()

print(band_powers.keys())  # 查看可用的频带名称

# 假设 samples 是 (1772, 3) 的坐标数据，band_powers['Delta'] 为功率值
x = samples[:, 0]
y = samples[:, 1]
z = samples[:, 2]
values = band_powers['Delta']  # 使用 Delta 频带的功率值

# 创建规则网格（3D 网格），这里选择较高的分辨率（100j）
grid_x, grid_y, grid_z = np.mgrid[min(x):max(x):100j, min(y):max(y):100j, min(z):max(z):100j]

# 使用 linear 插值进行插值（对于三维数据，'cubic' 不适用）
grid_values = griddata((x, y, z), values, (grid_x, grid_y, grid_z), method='linear')

# 计算有效点的数量（非 NaN 的点）
valid_points = np.count_nonzero(~np.isnan(grid_values))
print(f"有效点的数量: {valid_points}")

# 可视化结果（3D 可视化）
fig = plt.figure(figsize=(10, 8))#figsize=(10, 8)
ax = fig.add_subplot(111, projection='3d')

# 可视化 3D 散点图
ax.scatter(x, y, z, c=values, cmap='turbo', label='Data Points')

# 绘制插值后的数据（这里只是一个简单的展示方法）
# 这里只展示第一层的插值结果
ax.plot_surface(grid_x[:, :, 0], grid_y[:, :, 0], grid_values[:, :, 0], cmap='turbo', alpha=0.6)

plt.colorbar(ax.scatter(x, y, z, c=values, cmap='turbo'), ax=ax)
plt.title('3D Linear Interpolation of Delta Band Power')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import griddata
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler

def load_data():
    coords = np.load('coords.npy')
    band_powers = np.load('band_powers.npy', allow_pickle=True).item()
    return coords, band_powers

def interpolate_smooth(coords, values, method='linear'):
    grid_x = np.linspace(coords[:, 0].min(), coords[:, 0].max(), len(coords))
    grid_y = np.linspace(coords[:, 1].min(), coords[:, 1].max(), len(coords))
    grid_z = np.linspace(coords[:, 2].min(), coords[:, 2].max(), len(coords))
    grid_coords = np.column_stack([
        np.random.choice(grid_x, size=len(coords)),  
        np.random.choice(grid_y, size=len(coords)),  
        np.random.choice(grid_z, size=len(coords))
    ])
    interp_values = griddata(coords, values, grid_coords, method=method)
    return grid_coords, interp_values

def plot_3d_data(coords, values, interp_coords, interp_values, band):
    fig = plt.figure(figsize=(12, 5))
    vmin = min(values.min(), interp_values.min())
    vmax = max(values.max(), interp_values.max())
    
    ax1 = fig.add_subplot(121, projection='3d')
    scatter1 = ax1.scatter(coords[:, 0], coords[:, 1], coords[:, 2], c=values, cmap='jet', marker='o', vmin=vmin, vmax=vmax)
    ax1.set_title(f"Original SEEG Data - {band}")
    fig.colorbar(scatter1, ax=ax1, shrink=0.5, aspect=10)
    
    ax2 = fig.add_subplot(122, projection='3d')
    scatter2 = ax2.scatter(interp_coords[:, 0], interp_coords[:, 1], interp_coords[:, 2], c=interp_values, cmap='jet', marker='o', alpha=0.8, vmin=vmin, vmax=vmax)
    ax2.set_title(f"Interpolated SEEG Data - {band}")
    fig.colorbar(scatter2, ax=ax2, shrink=0.5, aspect=10)
    
    plt.tight_layout()
    plt.show()

def linear_interpolation(train_coords, train_values, test_coords, method='linear'):
    return griddata(train_coords, train_values, test_coords, method=method)

def calculate_metrics(true_values, pred_values):
    mask = ~np.isnan(pred_values)
    mse = mean_squared_error(true_values[mask], pred_values[mask])
    mae = mean_absolute_error(true_values[mask], pred_values[mask])
    rmse = np.sqrt(mse)
    r2 = r2_score(true_values[mask], pred_values[mask])
    return {"MSE": mse, "MAE": mae, "RMSE": rmse, "R2": r2}

def plot_scatter(metric_values, metric_name, validation_type, band):
    plt.figure(figsize=(8, 6))

    
    x_positions = np.arange(len(metric_values)) + np.random.uniform(-0.1, 0.1, len(metric_values))

    plt.scatter(x_positions, metric_values, alpha=0.7, color='b', edgecolors='black', s=100)  # 增大点的大小
    plt.xticks([]) 
    plt.xlabel("Data Points")
    plt.ylabel(metric_name)
    plt.title(f"{validation_type} Validation - {band} - {metric_name}")
    plt.grid(True)
    plt.show()

def cross_validate(coords, values, k=10, validation_type="Internal", band="Unknown", interpolation_method='linear'):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    all_metrics = {"MSE": [], "MAE": [], "RMSE": [], "R2": []}
    
    scaler = StandardScaler()
    coords = scaler.fit_transform(coords)  

    for train_idx, test_idx in kf.split(coords):
        train_coords, test_coords = coords[train_idx], coords[test_idx]
        train_values, test_values = values[train_idx], values[test_idx]
        
        pred_values = linear_interpolation(train_coords, train_values, test_coords, method=interpolation_method)
        metrics = calculate_metrics(test_values, pred_values)
        
        for metric in all_metrics:
            all_metrics[metric].append(metrics[metric])
    
    print(f"{validation_type} Validation - {band} Metrics:")
    for metric_name, values in all_metrics.items():
        avg_value = np.mean(values)
        print(f"{metric_name}: {avg_value:.4f}")
        plot_scatter(values, metric_name, validation_type, band)

def main():
    coords, band_powers = load_data()
    band = 'Theta'
    if band not in band_powers:
        print(f"Band {band} not found in data.")
        return
    
    band_values = np.array(band_powers[band])
    interp_coords, interp_values = interpolate_smooth(coords, band_values, method='cubic')  # 使用立方插值
    plot_3d_data(coords, band_values, interp_coords, interp_values, band)
    
    cross_validate(interp_coords, interp_values, k=15, validation_type="Internal", band=band, interpolation_method='cubic')
    cross_validate(interp_coords, interp_values, k=15, validation_type="External", band=band, interpolation_method='cubic')

if __name__ == "__main__":
    main()
```
